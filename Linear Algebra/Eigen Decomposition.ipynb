# 特徵分解

## 向量在特徵空間中的拆解與重建

此筆記展示：
1. 矩陣特徵分解
2. 向量投影到特徵向量基底
3. 從特徵空間重建
4. 驗證線性變換的一致性

````python
import numpy as np

# 1. 定義矩陣 A
A = np.array([[2, 1], [1, 2]])

# 2. 特徵分解
eigvals, eigvector = np.linalg.eig(A)
V_inv = np.linalg.inv(eigvector)

print("A =\n", A)
print("\n特徵值 Λ =\n", np.diag(eigvals))
print("\n特徵向量 V =\n", eigvector)

Back_A = eigvector @ np.diag(eigvals) @ V_inv
print("\n透過特徵分解重建 A =\n", Back_A)
````

### 數學說明

給定矩陣：

$$
A=
\begin{bmatrix}
2 & 1\\
1 & 2
\end{bmatrix}
$$

我們希望找到特徵分解：

$$
A = V\Lambda V^{-1}
$$

---

1. 求特徵值

$$
\det(A-\lambda I)=0
$$

$$
\det
\begin{bmatrix}
2-\lambda & 1\\
1 & 2-\lambda
\end{bmatrix}
=0
$$

$$
(2-\lambda)^2-1=0
$$

$$
\lambda^2-4\lambda+3=0
$$

$$
(\lambda-3)(\lambda-1)=0
$$

因此：

$$
\lambda_1=3,\qquad \lambda_2=1
$$

---

2. 求特徵向量

對應特徵向量可取：

$$
v_1=
\begin{bmatrix}
1\\
1
\end{bmatrix},
\qquad
v_2=
\begin{bmatrix}
1\\
-1
\end{bmatrix}
$$

組成矩陣：

$$
V=
\begin{bmatrix}
1 & 1\\
1 & -1
\end{bmatrix}
$$

$$
\Lambda=
\begin{bmatrix}
3 & 0\\
0 & 1
\end{bmatrix}
$$

---

3. 計算反矩陣

$$
V^{-1}
=
\frac{1}{-2}
\begin{bmatrix}
-1 & -1\\
-1 & 1
\end{bmatrix}
=
\begin{bmatrix}
\frac12 & \frac12\\
\frac12 & -\frac12
\end{bmatrix}
$$

---

## 4. 重建矩陣

$$
A = V\Lambda V^{-1}
$$

先計算：

$$
V\Lambda
=
\begin{bmatrix}
1 & 1\\
1 & -1
\end{bmatrix}
\begin{bmatrix}
3 & 0\\
0 & 1
\end{bmatrix}
=
\begin{bmatrix}
3 & 1\\
3 & -1
\end{bmatrix}
$$

再乘：

$$
V\Lambda V^{-1}
=
\begin{bmatrix}
3 & 1\\
3 & -1
\end{bmatrix}
\begin{bmatrix}
\frac12 & \frac12\\
\frac12 & -\frac12
\end{bmatrix}
=
\begin{bmatrix}
2 & 1\\
1 & 2
\end{bmatrix}
$$

---

5. 對應程式碼

```python
Back_A = eigvector @ np.diag(eigvals) @ V_inv
```

其數學意義為：

$$
A = V \Lambda V^{-1}
$$

即利用特徵值與特徵向量將原矩陣完整重建。

````python
# 投影到特徵基底
c = V_inv @ A
print("\n在特徵基底下的座標 c =", c)

# 4. 重建
x_reconstructed = eigvector @ c
print("\n重建回原向量 =", x_reconstructed)

# # 誤差
# print("\n重建誤差 =", np.linalg.norm(x - x_reconstructed))

# # 5. 驗證 Ax
# Ax = A @ x

# Lambda = np.diag(eigvals)
# Lambda_c = Lambda @ c
# Ax_via_eig = V @ Lambda_c

# print("\nA x =", Ax)
# print("\n透過特徵分解計算的 A x =", Ax_via_eig)
# print("\nAx 誤差 =", np.linalg.norm(Ax - Ax_via_eig))
````

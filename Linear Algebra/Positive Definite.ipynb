# 正定矩陣（Positive Definite Matrix）

本 Notebook 整理前面關於正定矩陣的討論，並加入 Python / NumPy 程式範例，示範如何檢查一個矩陣是否為正定矩陣。

## 1. 定義

對於一個實對稱矩陣 $A \in \mathbb{R}^{n \times n}$，如果對所有非零向量 $x \in \mathbb{R}^n$，都有

$$
x^T A x > 0
$$

則稱 $A$ 為**正定矩陣**。

其中 $x^T A x$ 稱為二次型。直觀上，正定矩陣表示：不論選擇哪個非零方向，這個二次型的值都嚴格大於 0。

## 2. 常用判別方式

對於實對稱矩陣 $A$，以下條件彼此等價：

1. 對所有非零向量 $x$，都有 $x^T A x > 0$。
2. 所有特徵值都大於 0。
3. 所有順序主子式（leading principal minors）都大於 0。
4. 可以做 Cholesky 分解：$A = LL^T$，且 $L$ 的對角線元素為正。

```python
import numpy as np


def is_symmetric(A, tol=1e-10):
    """檢查矩陣是否近似對稱。"""
    return np.allclose(A, A.T, atol=tol)


def is_positive_definite_by_eigenvalues(A, tol=1e-10):
    """使用特徵值檢查正定性。"""
    if not is_symmetric(A, tol):
        return False
    eigenvalues = np.linalg.eigvalsh(A)  # 對稱矩陣建議用 eigvalsh
    return np.all(eigenvalues > tol)


def is_positive_definite_by_cholesky(A):
    """使用 Cholesky 分解檢查正定性。"""
    try:
        np.linalg.cholesky(A)
        return True
    except np.linalg.LinAlgError:
        return False


def report_matrix(A, name="A"):
    """列印矩陣、特徵值與正定判斷。"""
    print(f"{name} =")
    print(A)
    print("對稱：", is_symmetric(A))
    if is_symmetric(A):
        print("特徵值：", np.linalg.eigvalsh(A))
    else:
        print("特徵值：", np.linalg.eigvals(A))
    print("用特徵值判斷是否正定：", is_positive_definite_by_eigenvalues(A))
    print("用 Cholesky 判斷是否正定：", is_positive_definite_by_cholesky(A))
```

## 3. 範例一：對角矩陣

矩陣

$$
A = 
\begin{bmatrix}
4 & 0 & 0 \\
0 & 1 & 0 \\
0 & 0 & 7
\end{bmatrix}
$$

是正定矩陣，因為它的特徵值就是對角元素 $4,1,7$，全部大於 0。

```python
A1 = np.array([[4, 0, 0], [0, 1, 0], [0, 0, 7]], dtype=float)

report_matrix(A1, "A1")
```

## 4. 範例二：非對角但對稱的正定矩陣

矩陣

$$
A =
\begin{bmatrix}
2 & 1 \\
1 & 2
\end{bmatrix}
$$

其特徵值為 $1$ 與 $3$，因此是正定矩陣。

也可以從二次型看出：

$$
x^T A x = 2x_1^2 + 2x_2^2 + 2x_1x_2
= (x_1+x_2)^2 + x_1^2 + x_2^2 > 0
$$

只要 $x \neq 0$，上式一定大於 0。

```python
A2 = np.array([[2, 1], [1, 2]], dtype=float)

report_matrix(A2, "A2")
```

## 5. 範例三：用主子式判斷

矩陣

$$
A =
\begin{bmatrix}
3 & -1 \\
-1 & 3
\end{bmatrix}
$$

順序主子式為：

$$
3 > 0
$$

以及

$$
\det(A) = 3 \cdot 3 - (-1)^2 = 8 > 0
$$

因此 $A$ 是正定矩陣。

```python
A3 = np.array([[3, -1], [-1, 3]], dtype=float)

report_matrix(A3, "A3")

print("第一個順序主子式：", A3[0, 0])
print("第二個順序主子式 det(A3)：", np.linalg.det(A3))
```

## 6. 範例四：3×3 正定矩陣

矩陣

$$
A =
\begin{bmatrix}
2 & -1 & 0 \\
-1 & 2 & -1 \\
0 & -1 & 2
\end{bmatrix}
$$

是常見的離散拉普拉斯型矩陣。它的特徵值都大於 0，因此是正定矩陣。

```python
A4 = np.array([[2, -1, 0], [-1, 2, -1], [0, -1, 2]], dtype=float)

report_matrix(A4, "A4")
```

## 7. 範例五：看起來較複雜但仍正定

矩陣

$$
A =
\begin{bmatrix}
5 & 2 \\
2 & 2
\end{bmatrix}
$$

第一個順序主子式為 $5>0$，行列式為

$$
5 \cdot 2 - 2 \cdot 2 = 6 > 0
$$

所以它是正定矩陣。

```python
A5 = np.array([[5, 2], [2, 2]], dtype=float)

report_matrix(A5, "A5")

print("第一個順序主子式：", A5[0, 0])
print("第二個順序主子式 det(A5)：", np.linalg.det(A5))
```

## 8. 反例一：正半定但不是正定

矩陣

$$
A =
\begin{bmatrix}
1 & 0 \\
0 & 0
\end{bmatrix}
$$

特徵值為 $1$ 與 $0$。因為其中一個特徵值等於 0，所以它是正半定矩陣，但不是正定矩陣。

```python
B1 = np.array([[1, 0], [0, 0]], dtype=float)

report_matrix(B1, "B1")
```

## 9. 反例二：不定矩陣

矩陣

$$
A =
\begin{bmatrix}
1 & 0 \\
0 & -1
\end{bmatrix}
$$

特徵值為 $1$ 與 $-1$，有正有負，因此是不定矩陣。

```python
B2 = np.array([[1, 0], [0, -1]], dtype=float)

report_matrix(B2, "B2")
```

## 10. 隨機產生正定矩陣

一個常見方法是先產生任意矩陣 $M$，再令

$$
A = M^T M + \epsilon I
$$

其中 $\epsilon > 0$。

$M^T M$ 一定是正半定矩陣；加上 $\epsilon I$ 後，會使所有特徵值至少增加 $\epsilon$，因此變成正定矩陣。

```python
np.random.seed(0)
M = np.random.randn(4, 4)
epsilon = 0.1
A_random = M.T @ M + epsilon * np.eye(4)

report_matrix(A_random, "A_random")
```

## 11. 小結

實務上常見的判斷方式如下：

- 小型矩陣：可用順序主子式判斷。
- 對稱矩陣：可直接檢查所有特徵值是否大於 0。
- 數值計算：常用 Cholesky 分解檢查正定性。

注意：正定矩陣通常預設討論的是實對稱矩陣；若矩陣不是對稱矩陣，正定性的定義與判斷需要更謹慎處理。

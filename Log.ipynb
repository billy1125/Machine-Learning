# 用 Python 介紹什麼是對數

這份 Notebook 用 Python 說明「對數」的基本概念、常見用法，以及為什麼對數常用於金融資料與乘法累積問題。

## 1. 對數的基本意思

對數可以理解成：

> 要把某個底數乘自己幾次，才會得到目標數字？

例如：

$$2^3 = 8$$

所以：

$$\log_2(8)=3$$

意思是：2 要乘自己 3 次，才會得到 8。

一般形式是：

$$\log_b(x)=y \Longleftrightarrow b^y=x$$

```python
import math

result = math.log(8, 2)
print(result)
```

上面的輸出是 `3.0`，表示：

$$2^3 = 8$$

也可以用 Python 驗證：

```python
print(2 ** result)
```

## 2. Python 裡常見的對數

Python 的 `math` 模組提供多種對數函數。

### 2.1 自然對數

`math.log(x)` 預設是自然對數，底數是 $e$，約等於 2.718。

$$\ln(x)=\log_e(x)$$

```python
x = 10
print(math.log(x))
```

### 2.2 以 10 為底的對數

`math.log10(x)` 是以 10 為底的對數。

```python
print(math.log10(1000))
```

輸出是 `3.0`，因為：

$$10^3=1000$$

### 2.3 以 2 為底的對數

`math.log2(x)` 是以 2 為底的對數。

```python
print(math.log2(8))
```

輸出是 `3.0`，因為：

$$2^3=8$$

## 3. 用表格觀察 log2 的結果

下面列出幾個數字，以及它們的 $\log_2(x)$。

```python
numbers = [1, 2, 4, 8, 16, 32]

for x in numbers:
    print(f"x = {x}, log2(x) = {math.log2(x)}")
```

可以看到：

$$2^0=1$$
$$2^1=2$$
$$2^2=4$$
$$2^3=8$$
$$2^4=16$$
$$2^5=32$$

所以 `log2(x)` 回答的是：

> 2 要乘自己幾次，才會得到 x？

## 4. 對數可以壓縮很大的數字

對數常用在數字變化很大的情況。例如：

```python
numbers = [1, 10, 100, 1000, 10000]

for x in numbers:
    print(f"x = {x}, log10(x) = {math.log10(x)}")
```

原本的數字差距很大：

$$1, 10, 100, 1000, 10000$$

取對數後變成：

$$0, 1, 2, 3, 4$$

因此，對數可以把快速變大的數字壓縮成比較容易比較的尺度。

## 5. 對數可以把乘法變成加法

這是對數非常重要的性質：

$$\log(a \times b)=\log(a)+\log(b)$$

這個性質在金融資料、成長率、比例變化中特別常用。

```python
day1 = 1.10  # 第一天上漲 10%
day2 = 1.20  # 第二天上漲 20%

total_growth = day1 * day2
log_total_growth = math.log(total_growth)

log_day1 = math.log(day1)
log_day2 = math.log(day2)

print("total_growth:", total_growth)
print("log(day1 * day2):", log_total_growth)
print("log(day1) + log(day2):", log_day1 + log_day2)
```

你會看到：

$$\log(1.10 \times 1.20) = \log(1.10)+\log(1.20)$$

也就是說，對數可以把乘法累積轉成加法累積。

## 6. 股票報酬中的對數

假設股票從 100 元漲到 110 元。普通報酬率是：

$$\frac{110-100}{100}=0.1$$

也就是上漲 10%。

對數報酬率則是：

$$\log\left(\frac{110}{100}\right)=\log(1.1)$$

```python
old_price = 100
new_price = 110

simple_return = (new_price - old_price) / old_price
log_return = math.log(new_price / old_price)

print("Simple return:", simple_return)
print("Log return:", log_return)
```

`simple_return` 是普通報酬率，表示上漲 10%。

`log_return` 是對數報酬率，適合處理連續多期的乘法變化，因為它可以把每期的比例變化轉成可相加的數值。

## 7. 對數與對數常態分佈

如果一個變數 $X$ 是由很多比例因子相乘形成：

$$X = X_0 \times R_1 \times R_2 \times \cdots \times R_N$$

取對數後：

$$\log(X)=\log(X_0)+\log(R_1)+\log(R_2)+\cdots+\log(R_N)$$

原本的乘法結構，取對數後變成加法結構。

根據中心極限定理，很多隨機變數相加後會趨近常態分佈。因此 $\log(X)$ 可能接近常態分佈，而 $X$ 本身則可能接近對數常態分佈。

## 8. 用 Python 模擬對數常態分佈

下面模擬一個股票價格：每天的對數報酬服從常態分佈，價格則由每日比例變化連續相乘形成。

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

num_samples = 100_000
num_steps = 252
initial_price = 100

mu = 0.0003
sigma = 0.02

# 每天的對數報酬
log_returns = np.random.normal(
    loc=mu,
    scale=sigma,
    size=(num_samples, num_steps)
)

# 累積對數報酬
total_log_returns = log_returns.sum(axis=1)

# 最終價格
final_prices = initial_price * np.exp(total_log_returns)

# 最終價格取對數
log_final_prices = np.log(final_prices)

print("Final prices mean:", np.mean(final_prices))
print("Final prices median:", np.median(final_prices))
print("log(final prices) mean:", np.mean(log_final_prices))
print("log(final prices) std:", np.std(log_final_prices))
```

### 8.1 觀察原始價格分佈

`final_prices` 通常會呈現右偏分佈，這是對數常態分佈常見的形狀。

```python
plt.figure(figsize=(8, 5))
plt.hist(final_prices, bins=100, density=True, alpha=0.7)
plt.title("Distribution of Final Prices")
plt.xlabel("Final price")
plt.ylabel("Density")
plt.show()
```

### 8.2 觀察 log(X) 的分佈

`log_final_prices` 通常會比較接近常態分佈，也就是鐘形曲線。

```python
plt.figure(figsize=(8, 5))
plt.hist(log_final_prices, bins=100, density=True, alpha=0.7)
plt.title("Distribution of log(Final Prices)")
plt.xlabel("log(final price)")
plt.ylabel("Density")
plt.show()
```

## 9. 總結

對數是在問：底數要幾次方，才會得到某個數。

Python 中常見的對數函數包括：

- `math.log(x)`：自然對數
- `math.log10(x)`：以 10 為底的對數
- `math.log2(x)`：以 2 為底的對數

對數的重要用途包括：

- 壓縮很大的數字尺度
- 把乘法變成加法
- 分析百分比變化與比例因子
- 解釋為什麼某些資料會接近對數常態分佈
